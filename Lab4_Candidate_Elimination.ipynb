```python
# Candidate Elimination Algorithm
# EnjoySport Dataset

# Attributes:
# Sky, AirTemp, Humidity, Wind, Water, Forecast

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes']
]


# Find-S: Most Specific Hypothesis
def find_s(data):
    hypothesis = None

    for row in data:
        attributes = row[:-1]
        label = row[-1]

        if label == 'Yes':
            if hypothesis is None:
                hypothesis = attributes.copy()
            else:
                for i in range(len(hypothesis)):
                    if hypothesis[i] != attributes[i]:
                        hypothesis[i] = '?'

    return hypothesis


# Candidate Elimination
def candidate_elimination(data):
    n = len(data[0]) - 1

    # Most specific hypothesis
    S = ['Ø'] * n

    # Most general hypothesis
    G = [['?'] * n]

    for row in data:
        x = row[:-1]
        label = row[-1]

        if label == 'Yes':
            # Generalize S
            for i in range(n):
                if S[i] == 'Ø':
                    S[i] = x[i]
                elif S[i] != x[i]:
                    S[i] = '?'

            # Remove hypotheses from G inconsistent with positive example
            G = [
                g for g in G
                if all(g[i] == '?' or g[i] == x[i] for i in range(n))
            ]

        else:
            # Specialize G to exclude negative example
            new_G = []

            for g in G:
                if all(g[i] == '?' or g[i] == x[i] for i in range(n)):
                    for i in range(n):
                        if S[i] != '?' and S[i] != 'Ø' and S[i] != x[i]:
                            new_h = g.copy()
                            new_h[i] = S[i]
                            if new_h not in new_G:
                                new_G.append(new_h)
                else:
                    new_G.append(g)

            G = new_G

    return S, G


# Run algorithm
S, G = candidate_elimination(data)

print("Most Specific Hypothesis (S):")
print(S)

print("\nMost General Boundary (G):")
for hypothesis in G:
    print(hypothesis)

print("\nFinal Version Space:")
print("All hypotheses between S and G")
```
